```python
##1. Load and Explore the Data 
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
mpg = sns.load_dataset("mpg")
```

```python
mpg.head()
```

```python
mpg.info()
```

```python
mpg.describe()
```

```python
##2. Horsepower vs. Fuel Economy 
```

Question: Does the vehicle's horsepower have a relationship with its fuel economy?

```python
sns.scatterplot(data=mpg, x="horsepower", y="mpg")

plt.title("Horsepower vs. Fuel Economy")
plt.xlabel("Horsepower")
plt.ylabel("Miles Per Gallon (MPG)")
plt.show()
```

Observation: The scatter plot shows a generally negative relationship between horsepower and the fuel economy. Vehicles with higher horsepower tend to have lower MPG, while vehicles with lower horsepower generally achieve better fuel economy. 

```python
## 3. Fuel Economy by Country
```

```python
Question: Which countries of origin had the highest average fuel economy?
```

```python
mpg_by_origin = mpg.groupby("origin")["mpg"].mean().reset_index()

mpg_by_origin
```

```python
mpg_by_origin = mpg.groupby("origin")["mpg"].mean().reset_index()

mpg_by_origin.columns = ["Country of Origin", "Average MPG"]

mpg_by_origin
```

```python
sns.barplot(data=mpg_by_origin, x="Country of Origin", y="Average MPG")

plt.title("Average Fuel Economy by Country of Origin")
plt.xlabel("Country of Origin")
plt.ylabel("Average MPG")
plt.show()
```

Observation: The average fuel economy varies by the country of origin. THe results here show the differences in the typical fuel economy of vechicles produced in the United States, Europe, and Japan.

```python
## 4. Fuel Economy Over Time
```

```python
mpg_by_year = mpg.groupby("model_year")["mpg"].mean().reset_index()

mpg_by_year
```

```python
sns.lineplot(data=mpg_by_year, x="model_year", y="mpg", marker="o")

plt.title("Fuel Economy Over Time")
plt.xlabel("Model Year")
plt.ylabel("Average MPG")
plt.show()
```

Observation: The average fuel economy generally has increased over the model years in the dataset. This here indicates that vehicles in the later model years tended to achieve higher MPGs than vehicles from the eariler model years. 

```python
## 5. Vehicle Weight vs. Fuel Economy 
```

```python
Question: How does vehicle weight relate to fuel economy?
```

```python
sns.scatterplot(data=mpg, x="weight", y="mpg")

plt.title("Vehicle Weight vs. Fuel Economy")
plt.xlabel("Vehicle Weight")
plt.ylabel("Miles Per Gallon (MPG)")
plt.show()
```

Observation: The heavier vehicles generally have the lower fuel economy. This data shows that MPG tends to decrease as vehicle weight increases.

```python
## 6. Fuel Economy by Number of Cylinders 
```

```python
mpg_by_cylinders = mpg.groupby("cylinders")["mpg"].mean().reset_index()

mpg_by_cylinders
```

```python
sns.barplot(data=mpg_by_cylinders, x="cylinders", y="mpg")

plt.title("Average Fuel Economy by Number of Cylinders")
plt.xlabel("Number of Cylinders")
plt.ylabel("Average MPG")
plt.show()
```

Observation: The vehicles with fewer cylinders generally have a higher average fuel economy, while the vehicles with more cylinders generally have lower average MPG.
